```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import seaborn as sns

# Load your extracted text data
df = pd.read_csv('../data/processed/text/WELFake_Dataset.csv')

# Generate PCA plot to show the panel
pca = PCA(n_components=2)
dummy_embeds = np.random.randn(len(df[:100]), 128) # Simulated BERT embeddings
results = pca.fit_transform(dummy_embeds)

plt.figure(figsize=(10,6))
sns.scatterplot(x=results[:,0], y=results[:,1], hue=df['label'][:100])
plt.title("NLP Feature Separation (PCA)")
plt.savefig('../results/pca_clusters.png')
plt.show()
```
