```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os

def analyze_svd(image_path):
    # The '../' is essential because the notebook is in the 'notebooks' folder
    img = cv2.imread(image_path, 0)
    
    if img is None:
        print(f"❌ Error: Could not find image at {image_path}")
        return
        
    # Perform SVD
    U, S, Vt = np.linalg.svd(img, full_matrices=False)
    
    # Plotting
    plt.figure(figsize=(8, 5))
    plt.semilogy(S)
    plt.title('Singular Value Spectrum (Forensic Analysis)')
    plt.xlabel('Rank')
    plt.ylabel('Singular Value (Log Scale)')
    
    # Save the result 'up' one level into the results folder
    output_path = '../results/svd_spectrum.png'
    plt.savefig(output_path)
    plt.show()
    print(f"✅ Success! Diagram saved to {output_path}")

# Run the function
analyze_svd('../data/processed/videos/sample_frame.jpg')
```
